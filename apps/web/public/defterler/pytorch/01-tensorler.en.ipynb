```python
import torch, random, numpy as np
torch.manual_seed(0); random.seed(0); np.random.seed(0)          # reproducibility
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('PyTorch', torch.__version__, '| device:', device)
```

## 1. Creating tensors
A tensor is a multi-dimensional array of numbers. It has a shape (`shape`), a data type (`dtype`) and a device (`device`).

```python
x = torch.arange(12, dtype=torch.float32).reshape(3, 4)
print(x)
print(x.shape, x.dtype, x.device)
z = torch.zeros(2, 3); o = torch.ones(2, 3); r = torch.randn(2, 3)   # zeros, ones, normal distribution
print(torch.from_numpy(np.array([1, 2, 3])).dtype)                  # from NumPy: int64
```

## 2. Slicing
`x[row, col]`; a slice is a **view** (not a copy).

```python
print(x[0], x[:, 1], x[1:, 2:])
v = x[0]; v[0] = 100                                                 # a view: x changes too
print(x[0, 0])
```

## 3. Broadcasting
Dimensions are aligned **from the right**; each pair must be equal or one of them must be 1.

```python
a = torch.ones(8, 1, 6, 1); b = torch.ones(7, 1, 5)
print((a + b).shape)                                                 # torch.Size([8, 7, 6, 5])
pred, target = torch.randn(64), torch.randn(64, 1)
print((pred - target).shape, '← silent bug!')         # torch.Size([64, 64])
print((pred - target.squeeze(1)).shape)                              # the correct way: torch.Size([64])
```

## 4. matmul, view and reshape

```python
W = torch.randn(4, 2)
print((x @ W).shape)                                                 # (3,4) @ (4,2) → (3,2)
try:
    torch.randn(64, 784) @ torch.randn(128, 10)
except RuntimeError as e:
    print('RuntimeError:', e)
t = x.T
print(t.is_contiguous())
try:
    t.view(-1)
except RuntimeError as e:
    print('view:', str(e)[:80])
print(t.reshape(-1).shape, t.contiguous().view(-1).shape)
```

## 5. Device
The model and the data must be on the same device.

```python
x = x.to(device)
print(x.device)
print((x * 2).sum().item())                                          # .item(): a Python number
```
