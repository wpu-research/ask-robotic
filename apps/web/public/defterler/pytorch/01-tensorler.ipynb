```python
import torch, random, numpy as np
torch.manual_seed(0); random.seed(0); np.random.seed(0)          # tekrarlanabilirlik
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('PyTorch', torch.__version__, '| device:', device)
```

## 1. Tensör oluşturma
Tensör, çok boyutlu bir sayı dizisidir. Şekli (`shape`), veri tipi (`dtype`) ve cihazı (`device`) vardır.

```python
x = torch.arange(12, dtype=torch.float32).reshape(3, 4)
print(x)
print(x.shape, x.dtype, x.device)
z = torch.zeros(2, 3); o = torch.ones(2, 3); r = torch.randn(2, 3)   # sıfırlar, birler, normal dağılım
print(torch.from_numpy(np.array([1, 2, 3])).dtype)                  # NumPy'dan: int64
```

## 2. Dilimleme
`x[satır, sütun]`; dilim bir **görünümdür** (kopya değildir).

```python
print(x[0], x[:, 1], x[1:, 2:])
v = x[0]; v[0] = 100                                                 # görünüm: x de değişir
print(x[0, 0])
```

## 3. Yayınlama
Boyutlar **sağdan** hizalanır; her çift eşit olmalı ya da biri 1 olmalı.

```python
a = torch.ones(8, 1, 6, 1); b = torch.ones(7, 1, 5)
print((a + b).shape)                                                 # torch.Size([8, 7, 6, 5])
pred, target = torch.randn(64), torch.randn(64, 1)
print((pred - target).shape, '← sessiz hata!')         # torch.Size([64, 64])
print((pred - target.squeeze(1)).shape)                              # doğrusu: torch.Size([64])
```

## 4. matmul, view ve reshape

```python
W = torch.randn(4, 2)
print((x @ W).shape)                                                 # (3,4) @ (4,2) → (3,2)
try:
    torch.randn(64, 784) @ torch.randn(128, 10)
except RuntimeError as e:
    print('RuntimeError:', e)
t = x.T
print(t.is_contiguous())
try:
    t.view(-1)
except RuntimeError as e:
    print('view:', str(e)[:80])
print(t.reshape(-1).shape, t.contiguous().view(-1).shape)
```

## 5. Cihaz
Model ve veri aynı cihazda olmalı.

```python
x = x.to(device)
print(x.device)
print((x * 2).sum().item())                                          # .item(): Python sayısı
```
